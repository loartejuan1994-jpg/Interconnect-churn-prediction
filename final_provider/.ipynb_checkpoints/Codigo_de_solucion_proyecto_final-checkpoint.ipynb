## <span style='color:#2E86AB;'> 2.1. Importar librerías </span>

````python
# Manejo de datos
import pandas as pd
import numpy as np
import math
import warnings

warnings.filterwarnings ('ignore')             # Maneja advertencias de pandas


# Librerías de visualización
import matplotlib.pyplot as plt
import seaborn as sns

# Procesamiento y modelado
from sklearn.preprocessing import StandardScaler, OrdinalEncoder           # Escala de variables numéricas
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, cross_val_score         # Dividir train/test/valid,  Búsqueda automática de hiperparámetros
from sklearn.linear_model import LogisticRegression                        # Modelo Regresión logística
from sklearn.ensemble import RandomForestClassifier                        # Modelo Bosque Aleatorio
from sklearn.tree import DecisionTreeClassifier                            # Modelo árbol de decisión
from sklearn.metrics import roc_auc_score, accuracy_score                  # que tan bien distinge entre las clases y % de predicciones correctas del total
from sklearn.metrics import roc_curve                                      # Curva para graficar el roc_auc
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay       # Matríz de confusión para ver donde falla el modelo

# Modelos de Boosting
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# Configuración visual
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

RANDOM_STATE = 12345                                                       # semilla fija para reproducibilidad en todo el notebook

print('Librerías importadas exitosamente')
````

# <span id="preprocesamiento" style='color:#1B3A6B;'> 3. PREPROCESAMIENTO </span>
---

## <span style='color:#2E86AB;'> 3.1. Unión de las 4 tablas </span>

Unir por la columna customerID usando  `left join` desde la tabla `contract` ya que esta tabla contiene a todos los clientes.

````python
# Unir 'contract' (tabla base) con internet, phone y personal usando 'customerID' como llave

df = df_contract.merge(df_personal, on= 'customerID', how= 'left')
df = df.merge(df_phone, on= 'customerID', how= 'left')                    # how='left' asegura que conservamos a TODOS los clientes de contract, aunque no tengan
df = df.merge(df_internet, on= 'customerID', how= 'left')

print(f'{"="*60}')
print(f"Dataset unificado: {df.shape[0]} filas, {df.shape[1]} columnas")
print(f'{"="*60}')
display(df.head(10))
````

## <span style='color:#2E86AB;'> 3.2. Construcción de la variable objetivo (**target**) </span>

La columna `EndDate` es `'No'` si el cliente sigue activo o tienen una fecha si canceló el cliente. Convertimos esto en una variable binaria **`target`**.

* `target = 1` → el cliente canceló (`EnDate` distinto de `No`)
* `target = 0` → el cliente sigue activo (`EndDate == NO` )

````python
# Construcción del target: 1 = canceló, 0 = activo
df['target'] = (df['EndDate'] != 'No').astype(int)

# Verificamos el balance de clases (importante para saber si necesitaremos
# técnicas de manejo de desbalance más adelante

conteo     = df['target'].value_counts() 
porcentaje = df['target'].value_counts(normalize= True) * 100

print(f'{"="*60}')
print('Distribución del target:')
print(f"Activos    (0) : {conteo[0]} ({porcentaje[0]:.1f} %)")
print(f"Cancelados (1) : {conteo[1]} ({porcentaje[1]:.1f} %)")
print(f'{"="*60}')
````

````python
# Guardamos columnas derivadas de EndDate para poder EXCLUIRLAS del análisis de churn
# por categoría más adelante (evita "fuga de información" / conclusiones contaminadas).
# BeginDate y EndDate no deben usarse como features del modelo directamente en su forma
# original (fecha), pero sí podemos extraer de ellas una feature útil y no contaminada:
# la antigüedad del cliente (en meses), calculada respecto a la fecha de referencia
# del proyecto (1 de febrero de 2020) para los clientes que siguen activos.

FECHA_REFERENCIA = pd.Timestamp('2020-02-01')

df['BeginDate'] = pd.to_datetime(df['BeginDate'])

# Para calcular antiguedad, si el cliente canceló usamos su EndDate real;
# si sigue activo, usamos la fecha de referencia del proyecto.
fecha_fin = pd.to_datetime(df['EndDate'], errors= 'coerce')              # errors='coerce' convierte valores que no se pueden transformar (Not para los 'No')
fecha_fin = fecha_fin.fillna(FECHA_REFERENCIA)                           # Si 'No' se convirtió en algo "vacío" (nulo), entonces fillna() rellena con (2020-02-01)

df['antiguedad_meses'] = ((fecha_fin - df['BeginDate']) / np.timedelta64(1, 'M')).round(1)    # Convierte en meses

print(f'{"="*60}')
print('Antiguedad calculada (en meses) - primeras filas:')
print(f'{"="*60}')

display(df[['BeginDate', 'EndDate', 'antiguedad_meses', 'target']].head())
````

## <span style='color:#2E86AB;'> 3.3. Tratamiento de nulos </span>

**Nota:** 

- La columna `totalcharges` viene como texto y algunos valores son espacios en blanco, se detecto al realizar las figuras de boxplots, anteriormente en el punto de tratamiento de nulos no se identificó estos valores dando inconsistencia al momento de graficar, esto ocurre en clientes muy nuevos con 0 meses de antigüedad o aún sin facturación total.

- Existen nulos en el dataset de clientes de las tablas (`Phone` y `Internet`), estos valores no deben ser imputados o remplazados con la media; por lo tanto serán remplazados con 'Sin Servicio' o 0, según el tipo de columna.

````python
# Nulos "reales" por inconsistencias de formato 'totalcharges' como texto con espacios.
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

print(f"Nulos generados en TotalCharges tras conversión: {df['TotalCharges'].isna().sum()}")

# Estos nulos corresponden a clientes nuevos: imputamos con 0, ya que aún no han
# acumulado cargos totales (es coherente con monthlycharges y antiguedad_meses ~ 0).

df['TotalCharges'] = df['TotalCharges'].fillna(0)
````

````python
# Convertir nombres de columnas en minúscula 
df.columns = df.columns.str.lower().str.strip()

# Valores nulos por columna
print(f'{"="*60}')
print('Valores antes de remplazar:')
print(f'{"="*60}')
display(df.isnull().sum())


# Rellenamos con 'Sin servicio': el propio NaN es informativo (cliente sin ese servicio),
# así que lo convertimos en una categoría explícita en vez de imputar con la moda.
columnas_phone    = ['multiplelines'] 
columnas_internet = ['internetservice', 'onlinesecurity', 'onlinebackup', 'deviceprotection', 'techsupport', 'streamingtv','streamingmovies']

for col in columnas_internet + columnas_phone:
    df[col] = df[col].fillna('Sin servicio')

print(f'{"="*60}')
print('Nulos restantes por columna:')
print(f'{"="*60}')

print(df.isna().sum()[df.isna().sum() > 0])
````

## <span style='color: #2E86AB;'> 3.4. Codificación de variables categóricas </span>

* Variables **binarias** (Sí/No) → mapeo directo a 0/1.
* Variables **categóricas con más de 2 categorías** → One-Hot Encoding, ya que no tienen un orden natural (por ejemplo: PaymentMethod, InternetService).

````python
# Variables binarias (Sí/No) presentes en el dataset
columnas_binarias = ['paperlessbilling', 'partner', 'dependents']
for col in columnas_binarias:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

# gender es binaria pero no Sí/No
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

# MultipleLines y las columnas tipo internet tienen 3 categorías ('Yes', 'No', 'Sin servicio')
# -> se manejarán con One-Hot Encoding junto con el resto de categóricas nominales.

columnas_ohe = ['type', 'paymentmethod', 'internetservice', 'multiplelines',
                'onlinesecurity', 'onlinebackup', 'deviceprotection',
                'techsupport', 'streamingtv', 'streamingmovies']

df_encoded = pd.get_dummies(df, columns= columnas_ohe, drop_first= True)              # Convierte variables categóricas a numéricas (One-Hot Encoding

print(f'{"="*60}')
print(f"Dataset codificado: {df_encoded.shape[0]} filas, {df_encoded.shape[1]} columnas")
print(f'{"="*60}')

display(df_encoded.head())
````

# <span id= "preparación" style='color:#1B3A6B;'> 5 PREPARACION PARA EL MODELO </span>
---

## <span style='color:#2E86AB;'> 5.1. División en conjuntos de entrenamiento y prueba </span>

Usamos `stratify=y` para conservar la misma proporción de clases (churn/no churn) en ambos conjuntos, dado el desbalance detectado en el EDA.

````python
# Creación de features y target
X = df_modelo.drop(columns=['target'])
y = df_modelo['target']

# Divición en train y test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size    = 0.25,
    random_state = RANDOM_STATE,
    stratify     = y                      # mantiene la proporción de clases en train y test
)

print(f'{"="*60}')
print(f"Train: {X_train.shape[0]} filas | Test: {X_test.shape[0]} filas")
print(f"Proporción de churn en train: {y_train.mean():.3f}")
print(f"Proporción de churn en test : {y_test.mean():.3f}")
print(f'{"="*60}')
````

````python
# Escalado de variables numéricas
columnas_a_escalar = ['monthlycharges', 'totalcharges', 'antiguedad_meses']


scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[columnas_a_escalar] = scaler.fit_transform(X_train[columnas_a_escalar])
X_test_scaled[columnas_a_escalar] = scaler.transform(X_test[columnas_a_escalar])
````

**Nota:**

El escalado es necesario para los modelos de Regresión Logística (sensible a escala), sin embargo no es estrictamente necesario para modelos basados en árboles (Random Forest, LightGBM, CatBoost), pero no les afecta negativamente, así que preparamos una versión escalada para reutilizar en todos los modelos lineales

---
# <span id="entrenamiento" style='color:#1B3A6B;'> 6. ENTRENAMIENTO DE MODELOS BASE (VALIDACIÓN CRUZADA) </span>
---

Comparamos 3 modelos base usando validación cruzada estratificada (5 folds) sobre el conjunto de entrenamiento, optimizando por AUC-ROC.

````python
# StratifiedKFold: garantiza que cada uno de los 5 folds tenga aproximadamente la misma proporción 73.5% / 26.5% 
cv = StratifiedKFold(n_splits= 5, shuffle= True, random_state= RANDOM_STATE )

modelos_base= {
    'Logistic Regression' : LogisticRegression(random_state= RANDOM_STATE, max_iter= 1000),
    'Random Forest'       : RandomForestClassifier(random_state= RANDOM_STATE, n_jobs= -1),
    'LightGBM'            : LGBMClassifier(random_state= RANDOM_STATE, verbose= -1),
    'CatBoost'            : CatBoostClassifier(random_state= RANDOM_STATE, verbose= 0)
}

resultados_base = {}

for nombre, modelo in modelos_base.items():
    # Regresión Logística usa datos escalados; los modelos de árboles usan datos sin escalar
    datos_train = X_train_scaled if nombre == 'Logistic Regression' else X_train

    # entrena y evalúa el modelo usando validación cruzada con 5 folds estratificados (misma proporción de clases),
    # calculando la métrica de evaluación en cada fold, y guarda las puntuaciones obtenidas en la variable scores
    scores = cross_val_score(modelo, datos_train, y_train, cv=cv, scoring='roc_auc', n_jobs=-1)
    resultados_base[nombre] = scores

    print(f"{nombre:20s} -> AUC-ROC promedio: {scores.mean():.4f} (+/- {scores.std():.4f})")
````

````python
# Visualizamos la comparación de modelos base
fig, ax = plt.subplots(figsize=(8, 5))
ax.boxplot(resultados_base.values(), labels = resultados_base.keys())
ax.set_ylabel('AUC-ROC (validación cruzada)')
ax.set_title('Fig. 5. Comparación de modelos base', fontweight='bold', y= 1.03)
plt.xticks(rotation=15)
plt.show()
````

> **Observaciones:**
>
> |Modelos|AUC-ROC promedio|
> |---|---|
> |CatBoost|0.8633|
> |LightGBM|0.8605|
> |Logistic Regression|0.8471|
> |Random Forest|0.8275|
> 
> Los modelos de boosting (LightGBM/CatBoost) superan a Regresión Logística y Random Forest en este tipo de problema tabular con variables
> categóricas y relaciones no lineales entre antigüedad, cargos y churn.

---
# <span id="hiperparámetros" style='color:#1B3A6B;'> 7. AJUSTE DE HIPERPARAMETROS Y MANEJO DE DESBALANCE DE CLASES </span>
---


|Modelo|AUC-ROC base|	Diferencia vs. CatBoost|
|---|---|---|
|CatBoost	|0.8633|	— |
|LightGBM	|0.8605	|-0.0028|
|Logistic Regression	|0.8471	|-0.0162|
|Random Forest	|0.8275	|-0.0358|

`LightGBM` quedó a solo 0.0028 de `CatBoost`, una diferencia que está dentro del rango de su propia desviación estándar (LightGBM: ±0.0096, CatBoost: ±0.0092). Eso significa que, con otra partición de datos o otra semilla aleatoria, LightGBM podría perfectamente haber quedado primero.


Seleccionamos los 2 mejores modelo bases y afinamos los hiperparámetros de cada uno, incorporando el desbalance de clases dectectadas en el EDA (26.5% churn). Comparamos los dos candidatos ya ajustados y seleccionamos el de mejor AUC-ROC.

* **CatBoost** → `scale_pos_weight` o auto_class_weights='Balanced'
* **LightGBM** → `class_weight='balanced'` o scale_pos_weight

````python
# DESBALANCE: Equilibra los conjuntos de datos desequilibrados en clasificación binaria
# Calculamos scale_pos_weight = (negativos / positivos) para los modelos de boosting
n_negativos = (y_train == 0).sum()
n_positivos = (y_train == 1).sum()
scale_pos_weight = (n_negativos / n_positivos)

print(f'{"="*60}')
print(f"scale_pos_weight calculado: {scale_pos_weight:.3f}")
print(f'{"="*60}')
````

````python
# Ajuste de hiperparámetros para CatBoost (mejor modelo base)
param_grid_catboost = {
    'iterations':[200,400,600],
    'depth'     :[4, 6, 8],
    'learning_rate' :[0.01, 0.05, 0.1],
    'l2_leaf_reg'   :[1, 3, 5, 7]                   # regularización L2, ayuda a evitar sobreajuste

}

catboost_balanced = CatBoostClassifier(
    random_state     = RANDOM_STATE,
    scale_pos_weight = scale_pos_weight,            # compensa el desbalance de clases (~26.5% churn)
    verbose          = 0,                           # silencia el log de entrenamiento de CatBoost
    eval_metric      = 'AUC'                        # CatBoost optimiza internamente por AUC
)

modelo_cat = GridSearchCV(
    estimator  = catboost_balanced,
    param_grid = param_grid_catboost,
    scoring    = 'roc_auc',
    cv         = cv,
    n_jobs     = -1,                                # nota: CatBoost + n_jobs=-1 puede ser lento por uso interno de CPU
    verbose    = 1                                  # Muestra el progreso del entrenamiento cada 1 iteración
)
modelo_cat.fit(X_train, y_train)

print(f'{"="*60}')
print('Mejores hiperparámetros encontrados (CatBoost):')
print(modelo_cat.best_params_)                                                 # Los mejores parámetros quedan guardado en modelo.best_params_.
print(f'Mejor AUC-ROC en validación cruzada: {modelo_cat.best_score_:.4f}')
print(f'{"="*60}')
````

````python
# Modelo LightGBM
param_grid_lgbm = {
    'n_estimators' : [100, 200, 400],
    'max_depth'    : [4, 6, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'num_leaves'   : [15, 31, 63]
}

lgbm_balanced = LGBMClassifier(
    random_state= RANDOM_STATE,
    class_weight= 'balanced',                     # compensa el desbalance de clases (~26.5% churn)
    verbose     = -1
)

modelo_lgbm = GridSearchCV(
    estimator = lgbm_balanced,
    param_grid= param_grid_lgbm,
    scoring   = 'roc_auc',
    cv        = cv,
    n_jobs    = -1,
    verbose   = 1
)
modelo_lgbm.fit(X_train, y_train)

print(f'{"="*60}')
print('=== LightGBM ===')
print('Mejores hiperparámetros:', modelo_lgbm.best_params_)
print()
print(f'Mejor AUC-ROC en validación cruzada: {modelo_lgbm.best_score_:.4f}')
print(f'{"="*60}')
````

````python
# Comparamos los dos candidatos ya ajustados y seleccionamos el de mejor AUC-ROC
candidatos_ajustados = {
    'CatBoost': modelo_cat,
    'LightGBM': modelo_lgbm
}

print(f'{"="*60}')
print('=== Comparación final tras ajuste de hiperparámetros ===')

for nombre, grid in candidatos_ajustados.items():
    print(f'{nombre:10s} -> AUC-ROC: {grid.best_score_:.4f}')

# Selección automática: el mejor_modelo queda definido según cuál grid tuvo el mayor
# best_score_, sin importar cuál familia de modelo haya ganado
nombre_ganador = max(candidatos_ajustados, key=lambda n: candidatos_ajustados[n].best_score_)
mejor_modelo = candidatos_ajustados[nombre_ganador].best_estimator_

print(f'\nModelo final seleccionado: {nombre_ganador}')
print(f'AUC-ROC (validación cruzada): {candidatos_ajustados[nombre_ganador].best_score_:.4f}')
print(f'{"="*60}')
````

````python
# Guardamos el mejor modelo encontrado para la evaluación final

print(f'{"="*60}')
print('Mejores parámetros del modelo CatBoost')
print(mejor_modelo.get_params())                  # Verificar los parámetros mejores parámetros
print(f'{"="*60}')
````

> **Nota:**
>
> * El modelo con mejor AUC-ROC en la comparación con hiperparámetros ajustados es CatBoost, aunque con una diferencia mínima frente a LightGBM. Antes de entrenarlo, calculamos scale_pos_weight = negativos / positivos para compensar el desbalance de clases detectado en el EDA (26.5% de clientes con churn frente a 73.5% activos), de modo que el modelo penalice más los errores sobre la clase minoritaria durante el entrenamiento.
>   
> * `GridSearchCV` guarda en `best_params_`, la combinación de hiperparámetros que obtuvo el mejor AUC-ROC promedio en la validación cruzada (depth=4, iterations=600, l2_leaf_reg=1, learning_rate=0.05). Con `refit= True` (valor por defecto), automáticamente reentrena un CatBoostClassifier con esos parámetros sobre el 100% del conjunto de entrenamiento, y guarda ese modelo ya entrenado en `best_estimator_`.
> 
>    Por eso `mejor_modelo = grid_search_cb.best_estimator_`, no es solo la configuración de parámetros; es el objeto del modelo completo y entrenado, listo para usarse directamente con `.predict()` y `.predict_proba()` sobre el conjunto de prueba, sin necesidad de ningún paso adicional de reentrenamiento.

# <span id="evaluación" style='color:#1B3A6B;'> 8. EVALUACION FINAL EN EL CONJUNTO DE PRUEBA </span>

````python
# Predicciones sobre el conjunto de prueba, nunca antes visto por el modelo
y_pred       = mejor_modelo.predict(X_test)
y_pred_proba = mejor_modelo.predict_proba(X_test)[:, 1]                    # (:) selecciona todas las filas de la matriz; (1) quédate solo con la columna en la posición 1

# Métrica principal: AUC-ROC
auc_roc_final = roc_auc_score(y_test, y_pred_proba)

# Métrica adicional: exactitud
exactitud_final = accuracy_score(y_test, y_pred)

print(f'{"="*60}')
print(f"AUC-ROC en test             : {auc_roc_final:.4f}")
print(f"Exactitud (acurracy) en test: {exactitud_final:.4f}")


# Ubicamos el resultado dentro de la tabla de puntuación del proyecto
if auc_roc_final < 0.75:
    sp = 0

elif auc_roc_final < 0.81:
    sp = 4
elif auc_roc_final < 0.85:
    sp = 4.5
elif auc_roc_final < 0.87:
    sp = 5
elif auc_roc_final < 0.88:
    sp = 5.5
else:
    sp = 6

print(f"Puntuación estimada del proyecto (SP): {sp}")
print(f'{"="*60}')
````

````python
# Curva ROC

# La función entrega tres arreglos que necesitas para construir la gráfica:
# fpr (False Positive Rate): Eje X de la gráfica.
# tpr (True Positive Rate / Recall): Eje Y de la gráfica.
# thresholds: Los umbrales de decisión que el algoritmo evaluó para calcular cada punto de FPR y TPR.

fpr, tpr, thresholds= roc_curve(y_test, y_pred_proba)                  # usando las etiquetas reales (y_test) y las probabilidades predichas de la clase positiva (y_pred_proba)

plt.figure(figsize = (6, 6))
plt.plot(fpr, tpr, 
         label= f"AUC-ROC = {auc_roc_final:.4f}", 
         color='#DD8452')                                              # Dibuja la curva ROC usando fpr en el eje X, tpr en el eje Y

plt.plot([0, 1], [0, 1], 
         linestyle= '--', 
         color='gray', 
         label= 'Modelo aleatorio')                                   # Dibuja una línea diagonal punteada desde el punto (0,0) al (1,1)
plt.xlabel('Tasa de falsos positivos (FPR)')
plt.ylabel('Tasa de verdaderos positivos (TPR)')
plt.title('Fig. 6. Curva ROC - Modelo final', fontweight='bold', y= 1.03)
plt.legend(loc = 'lower right')
plt.show()


print(f'{"="*60}')

print(f"AUC-ROC en test             : {auc_roc_final:.4f} ({auc_roc_final:.2%})   ← métrica principal")
print(f"Exactitud (acurracy) en test: {exactitud_final:.4f} ({exactitud_final:.2%})   ← métrica secundaria")
print(f'{"="*60}')
````

> **Observaciones:**
>
> * En la figura 6, la línea curva de color 'naranja' representa el rendimiento del modelo CatBoost con `scale_pos_weight` en el conjunto de prueba.
>
> * La curva `AUC-ROC` muestra un **excelente rendimiento** (86.7%), acercándose a la esquina superior izquierda, esto indica una alta capacidad de distingir entre clientes que abandonan y los que no.
>
> **Interpretación Técnica de la curva**
>
> * La curva se encuentra por encima de la línea diagonal gris (modelo aleatorio).
> * Esto confirma que el modelo tiene una capacidad discriminativa superior al azar para identificar clientes en riesgo de abandono.
> * El `AUC-ROC` de 86.7% (métrica principal)clasifica como 'muy bueno' en la escala estándar de evaluación de modelos, con una exactitud (acurracy) de 77.80% (métrica secundaria) y con una puntuación estimada del proyecto **(SP= 5)**.

> **Observaciones:**
>
>```
>                             Predicho
>                      Activos        Cancelados
>  Real  Activos      [ ✅ TN           ❌ FP   ]   ← clientes Activos
>        Cancelados   [ ❌ FN           ✅ TP   ]   ← clientes Cancelados
>-----------------------------------------------------------------
>                             Predicho
>                      Activos        Cancelados
>  Real  Activos      [ 996           298   ]   = 1294 ✅
>        Cancelados   [  93           374   ]   =  467 ✅
>-----------------------------------------------------------------
>               ──────────────────────────────────
>    TOTAL PRED        1089           672       = 1761  total
>-----------------------------------------------------------------
>```
> **Interpretación de resultados:**
> 
> - **TN = 996:** Clientes reales `Activos`, predichos como `Activos`. 
>
> - **FP = 298:** Clientes reales `Activos`, predichos (por error) como `Cancelados`.             
>
> - **FN =  93:** Clientes reales `Cancelados`, predichos (por error) como `Activos`.   ← **(bajo, es buena señal)** 
>
> - **TP = 374:** Clientes reales `Cancelados`, predichos como `Cancelados`.
>
> **Análisis de eficiencia:**
> * **Recall (Sensibilidad):** clase Cancelado (374/467) = `80.1%` (el modelo detecta ~8 de cada 10 clientes que sí cancelan)
> * **Precisión:** (374/672) = `55.7%`, números coherentes con un AUC-ROC de ~0.86-0.87 en un problema con desbalance moderado.
