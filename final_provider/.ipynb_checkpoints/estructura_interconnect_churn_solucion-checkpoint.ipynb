# Proyecto Interconnect: Predicción de Cancelación de Clientes (Churn)
## Etapa 2 — Código de solución

**Objetivo:** construir un modelo de clasificación que prediga si un cliente cancelará su
servicio (`EndDate` ≠ `'No'`), maximizando el **AUC-ROC** (métrica principal) y reportando
la **exactitud** (métrica adicional).

**Criterios de evaluación (SP):**

| AUC-ROC | SP |
|---|---|
| < 0.75 | 0 |
| 0.75 – 0.81 | 4 |
| 0.81 – 0.85 | 4.5 |
| 0.85 – 0.87 | 5 |
| 0.87 – 0.88 | 5.5 |
| ≥ 0.88 | 6 |

**Contenido de este notebook:**
1. Preprocesamiento (unión de tablas, construcción del target, nulos, codificación)
2. Análisis exploratorio y de correlaciones
3. Modelos base con validación cruzada
4. Ajuste de hiperparámetros y manejo de desbalance de clases
5. Evaluación final en el conjunto de prueba
6. Conclusiones


## 0. Librerías necesarias

```
# --- Manejo de datos ---
import pandas as pd
import numpy as np

# --- Visualización ---
import matplotlib.pyplot as plt
import seaborn as sns

# --- Preprocesamiento y modelado ---
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, OrdinalEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, accuracy_score, roc_curve, confusion_matrix, ConfusionMatrixDisplay

# --- Modelos de boosting (si no están instalados, descomentar la siguiente línea) ---
# !pip install lightgbm catboost --quiet
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

import warnings
warnings.filterwarnings('ignore')

# Configuración visual
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

RANDOM_STATE = 12345  # semilla fija para reproducibilidad en todo el notebook
```

## 1. Carga de los datasets

```
# Cargamos los 4 archivos fuente. Cada uno aporta una parte de la información del cliente.
df_contract = pd.read_csv('contract.csv')   # información contractual (incluye EndDate -> target)
df_personal = pd.read_csv('personal.csv')   # datos demográficos del cliente
df_internet = pd.read_csv('internet.csv')   # servicios de internet contratados
df_phone    = pd.read_csv('phone.csv')      # servicios telefónicos contratados

datasets = {
    'contract': df_contract,
    'personal': df_personal,
    'internet': df_internet,
    'phone': df_phone
}

# Revisión rápida de dimensiones para confirmar que la carga fue correcta
for nombre, df in datasets.items():
    print(f'{nombre:10s} -> {df.shape[0]} filas, {df.shape[1]} columnas')
```

## 2. Preprocesamiento

### 2.1 Unión de las 4 tablas

Unimos todo por `customerID` usando `left join` desde `contract`, ya que esta tabla contiene
a **todos** los clientes (los que no tienen internet o teléfono simplemente no aparecerán en
esas tablas y quedarán como `NaN` tras el merge, lo cual es información válida: "no tiene ese servicio").

```
# Unimos contract (tabla base) con personal, internet y phone usando customerID como llave.
# how='left' asegura que conservamos a TODOS los clientes de contract, aunque no tengan
# internet o teléfono (esos casos quedarán como NaN y se tratarán en el siguiente paso).
df = df_contract.merge(df_personal, on='customerID', how='left')
df = df.merge(df_internet, on='customerID', how='left')
df = df.merge(df_phone, on='customerID', how='left')

print(f'Dataset unificado: {df.shape[0]} filas, {df.shape[1]} columnas')
df.head()
```

### 2.2 Construcción de la variable objetivo (target)

La columna `EndDate` vale `'No'` si el cliente sigue activo, o contiene una fecha si el
cliente canceló. Convertimos esto en una variable binaria `target`:
- `target = 1` → el cliente **canceló** (`EndDate` distinto de `'No'`)
- `target = 0` → el cliente **sigue activo** (`EndDate == 'No'`)

```
# Construcción del target: 1 = canceló, 0 = activo
df['target'] = (df['EndDate'] != 'No').astype(int)

# Verificamos el balance de clases (importante para saber si necesitaremos
# técnicas de manejo de desbalance más adelante)
conteo = df['target'].value_counts()
porcentaje = df['target'].value_counts(normalize=True) * 100

print('Distribución del target:')
print(f"  Activos (0)   : {conteo[0]} ({porcentaje[0]:.1f}%)")
print(f"  Cancelados (1): {conteo[1]} ({porcentaje[1]:.1f}%)")
```

```
# Guardamos columnas derivadas de EndDate para poder EXCLUIRLAS del análisis de churn
# por categoría más adelante (evita "fuga de información" / conclusiones contaminadas).
# BeginDate y EndDate no deben usarse como features del modelo directamente en su forma
# original (fecha), pero sí podemos extraer de ellas una feature útil y no contaminada:
# la antigüedad del cliente (en meses), calculada respecto a la fecha de referencia
# del proyecto (1 de febrero de 2020) para los clientes que siguen activos.

FECHA_REFERENCIA = pd.Timestamp('2020-02-01')

df['BeginDate'] = pd.to_datetime(df['BeginDate'])

# Para calcular antigüedad: si el cliente canceló, usamos su EndDate real;
# si sigue activo, usamos la fecha de referencia del proyecto.
fecha_fin = pd.to_datetime(df['EndDate'], errors='coerce')  # NaT para los 'No'
fecha_fin = fecha_fin.fillna(FECHA_REFERENCIA)

df['antiguedad_meses'] = ((fecha_fin - df['BeginDate']) / np.timedelta64(1, 'M')).round(1)

print('Antigüedad calculada (en meses) - primeras filas:')
df[['BeginDate', 'EndDate', 'antiguedad_meses', 'target']].head()
```

### 2.3 Tratamiento de nulos

Existen dos tipos de nulos en este dataset:
1. **Nulos "reales"** por inconsistencias de formato (ej. `TotalCharges` como texto con espacios).
2. **Nulos "estructurales"**: clientes que no tienen internet o teléfono, y por tanto no
   aparecen en esas tablas. Estos NO deben eliminarse ni imputarse con la media: se rellenan
   con una categoría explícita `'Sin servicio'` o con 0, según el tipo de columna.

```
# --- Caso 1: TotalCharges viene como texto y algunos valores son espacios en blanco ---
# (ocurre en clientes muy nuevos con 0 meses de antigüedad, aún sin facturación total)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

print(f"Nulos generados en TotalCharges tras conversión: {df['TotalCharges'].isna().sum()}")

# Estos nulos corresponden a clientes nuevos: imputamos con 0, ya que aún no han
# acumulado cargos totales (es coherente con MonthlyCharges y antigüedad_meses ~ 0).
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```
# --- Caso 2: columnas de servicios de internet/teléfono con NaN por ausencia de servicio ---
columnas_internet = ['InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
                      'TechSupport', 'StreamingTV', 'StreamingMovies']
columnas_phone = ['MultipleLines']

# Rellenamos con 'Sin servicio': el propio NaN es informativo (cliente sin ese servicio),
# así que lo convertimos en una categoría explícita en vez de imputar con la moda.
for col in columnas_internet + columnas_phone:
    df[col] = df[col].fillna('Sin servicio')

# Verificación final: no deben quedar nulos en el dataset
print('Nulos restantes por columna:')
print(df.isna().sum()[df.isna().sum() > 0])
```

### 2.4 Codificación de variables categóricas

- Variables **binarias** (Sí/No) → mapeo directo a 0/1.
- Variables **categóricas con más de 2 categorías** → One-Hot Encoding, ya que no tienen
  un orden natural (ej. `PaymentMethod`, `InternetService`).

```
# Variables binarias (Sí/No) presentes en el dataset
columnas_binarias = ['Partner', 'Dependents', 'PaperlessBilling']
for col in columnas_binarias:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

# gender es binaria pero no Sí/No
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

# MultipleLines y las columnas tipo internet tienen 3 categorías ('Yes', 'No', 'Sin servicio')
# -> se manejarán con One-Hot Encoding junto con el resto de categóricas nominales.
columnas_ohe = ['Type', 'PaymentMethod', 'InternetService', 'MultipleLines',
                'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
                'TechSupport', 'StreamingTV', 'StreamingMovies']

df_encoded = pd.get_dummies(df, columns=columnas_ohe, drop_first=True)

print(f'Dataset codificado: {df_encoded.shape[0]} filas, {df_encoded.shape[1]} columnas')
df_encoded.head()
```

## 3. Análisis exploratorio y de correlaciones

**Importante:** excluimos `BeginDate`, `EndDate` y `customerID` de cualquier análisis o
modelo, ya que `EndDate` es la fuente directa del target (usarla sería fuga de información)
y `customerID` no aporta información predictiva real.

```
# Columnas que deben excluirse del análisis de churn y del modelado
columnas_excluir = ['customerID', 'BeginDate', 'EndDate']

df_modelo = df_encoded.drop(columns=columnas_excluir)

print('Columnas excluidas (contaminadas o sin valor predictivo):', columnas_excluir)
print(f'Columnas disponibles para el análisis/modelo: {df_modelo.shape[1] - 1}')  # -1 por el target
```

### 3.1 Distribución del target

```
fig, ax = plt.subplots(figsize=(6, 4))
df['target'].value_counts().plot(kind='bar', color=['#4C72B0', '#DD8452'], ax=ax)
ax.set_xticklabels(['Activo (0)', 'Cancelado (1)'], rotation=0)
ax.set_ylabel('Número de clientes')
ax.set_title('Distribución de la variable objetivo (churn)')
plt.show()

# CONCLUSIÓN ESPERADA: se observará un desbalance de clases (los clientes activos suelen
# ser mayoría, ~73-74%, frente a ~26-27% de cancelados). Esto se debe tener en cuenta al
# entrenar los modelos (class_weight, scale_pos_weight, etc.).
```

### 3.2 Variables numéricas vs. churn (boxplots)

```
columnas_numericas = ['MonthlyCharges', 'TotalCharges', 'antiguedad_meses']

fig, axes = plt.subplots(1, len(columnas_numericas), figsize=(15, 5))
for ax, col in zip(axes, columnas_numericas):
    sns.boxplot(data=df, x='target', y=col, ax=ax, palette=['#4C72B0', '#DD8452'])
    ax.set_xticklabels(['Activo', 'Cancelado'])
    ax.set_title(col)
plt.tight_layout()
plt.show()

# CONCLUSIÓN ESPERADA:
# - antiguedad_meses: los clientes que cancelan suelen tener MENOR antigüedad
#   (el churn ocurre más en los primeros meses de contrato).
# - MonthlyCharges: los clientes que cancelan tienden a pagar cargos mensuales MÁS ALTOS.
# - TotalCharges: relacionado con antigüedad, los cancelados tienden a tener TotalCharges
#   más bajos (menos tiempo acumulando facturación).
```

### 3.3 Tasa de churn por variable categórica

```
columnas_categoricas_originales = ['Type', 'PaymentMethod', 'InternetService', 'SeniorCitizen',
                                    'Partner', 'Dependents', 'PaperlessBilling']

fig, axes = plt.subplots(4, 2, figsize=(14, 18))
axes = axes.flatten()

for i, col in enumerate(columnas_categoricas_originales):
    tasa_churn = df.groupby(col)['target'].mean().sort_values(ascending=False)
    tasa_churn.plot(kind='bar', ax=axes[i], color='#DD8452')
    axes[i].set_title(f'Tasa de churn por {col}')
    axes[i].set_ylabel('Tasa de churn')
    axes[i].tick_params(axis='x', rotation=30)

# Ocultamos el último subplot vacío (7 variables en una grilla de 8)
fig.delaxes(axes[-1])
plt.tight_layout()
plt.show()

# CONCLUSIÓN ESPERADA:
# - Type (tipo de contrato): los contratos 'Month-to-month' (mes a mes) presentan la tasa
#   de churn más alta; los contratos de 1-2 años son mucho más estables.
# - PaymentMethod: el pago con 'Electronic check' suele asociarse a mayor churn.
# - InternetService: la fibra óptica ('Fiber optic') suele tener mayor tasa de cancelación
#   que DSL (posiblemente por precio o problemas de servicio).
```

### 3.4 Matriz de correlación (variables numéricas y target)

```
columnas_corr = ['MonthlyCharges', 'TotalCharges', 'antiguedad_meses', 'target']

matriz_corr = df[columnas_corr].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(matriz_corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Matriz de correlación')
plt.show()

# CONCLUSIÓN ESPERADA: antiguedad_meses debería mostrar la correlación negativa más fuerte
# con el target (a mayor antigüedad, menor probabilidad de cancelar), mientras que
# MonthlyCharges debería mostrar una correlación positiva moderada.
```

## 4. Preparación para el modelado

### 4.1 División en conjuntos de entrenamiento y prueba

Usamos `stratify=y` para conservar la misma proporción de clases (churn/no churn) en ambos
conjuntos, dado el desbalance detectado en el EDA.

```
X = df_modelo.drop(columns=['target'])
y = df_modelo['target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=RANDOM_STATE,
    stratify=y  # mantiene la proporción de clases en train y test
)

print(f'Train: {X_train.shape[0]} filas | Test: {X_test.shape[0]} filas')
print(f'Proporción de churn en train: {y_train.mean():.3f}')
print(f'Proporción de churn en test : {y_test.mean():.3f}')
```

```
# Escalado de variables numéricas: necesario para Regresión Logística (sensible a escala),
# no es estrictamente necesario para modelos basados en árboles (Random Forest, LightGBM,
# CatBoost), pero no les afecta negativamente, así que preparamos una versión escalada
# para reutilizar en todos los modelos lineales.
columnas_a_escalar = ['MonthlyCharges', 'TotalCharges', 'antiguedad_meses']

scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[columnas_a_escalar] = scaler.fit_transform(X_train[columnas_a_escalar])
X_test_scaled[columnas_a_escalar] = scaler.transform(X_test[columnas_a_escalar])
```

## 5. Entrenamiento de modelos base (validación cruzada)

Comparamos 3 modelos base usando validación cruzada estratificada (5 folds) sobre el
conjunto de entrenamiento, optimizando por AUC-ROC.

```
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

modelos_base = {
    'Logistic Regression': LogisticRegression(random_state=RANDOM_STATE, max_iter=1000),
    'Random Forest': RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
    'LightGBM': LGBMClassifier(random_state=RANDOM_STATE, verbose=-1),
    'CatBoost': CatBoostClassifier(random_state=RANDOM_STATE, verbose=0)
}

resultados_base = {}

for nombre, modelo in modelos_base.items():
    # Regresión Logística usa datos escalados; los modelos de árboles usan datos sin escalar
    datos_train = X_train_scaled if nombre == 'Logistic Regression' else X_train

    scores = cross_val_score(modelo, datos_train, y_train, cv=cv, scoring='roc_auc', n_jobs=-1)
    resultados_base[nombre] = scores
    print(f'{nombre:22s} -> AUC-ROC promedio: {scores.mean():.4f} (+/- {scores.std():.4f})')
```

```
# Visualizamos la comparación de modelos base
fig, ax = plt.subplots(figsize=(8, 5))
ax.boxplot(resultados_base.values(), labels=resultados_base.keys())
ax.set_ylabel('AUC-ROC (validación cruzada)')
ax.set_title('Comparación de modelos base')
plt.xticks(rotation=15)
plt.show()

# CONCLUSIÓN ESPERADA: los modelos de boosting (LightGBM/CatBoost) normalmente superan
# a Regresión Logística y Random Forest en este tipo de problema tabular con variables
# categóricas y relaciones no lineales entre antigüedad, cargos y churn.
```

## 6. Ajuste de hiperparámetros y manejo del desbalance de clases

Seleccionamos el/los modelo(s) con mejor desempeño base y afinamos sus hiperparámetros,
incorporando el manejo del desbalance de clases detectado en el EDA (~26-27% de churn):

- **Logistic Regression / Random Forest** → `class_weight='balanced'`
- **LightGBM** → `class_weight='balanced'` o `scale_pos_weight`
- **CatBoost** → `scale_pos_weight` o `auto_class_weights='Balanced'`

```
# Calculamos scale_pos_weight = (negativos / positivos) para los modelos de boosting
n_negativos = (y_train == 0).sum()
n_positivos = (y_train == 1).sum()
scale_pos_weight = n_negativos / n_positivos

print(f'scale_pos_weight calculado: {scale_pos_weight:.3f}')
```

```
# --- Ejemplo de ajuste de hiperparámetros para LightGBM (asumiendo que fue el mejor base) ---
# NOTA: ajustar el grid según el/los modelo(s) que hayan mostrado mejor desempeño en la
# comparación anterior. Este bloque es un ejemplo reproducible.

param_grid_lgbm = {
    'n_estimators': [100, 200, 400],
    'max_depth': [4, 6, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'num_leaves': [15, 31, 63]
}

lgbm_balanced = LGBMClassifier(
    random_state=RANDOM_STATE,
    class_weight='balanced',  # compensa el desbalance de clases
    verbose=-1
)

grid_search = GridSearchCV(
    estimator=lgbm_balanced,
    param_grid=param_grid_lgbm,
    scoring='roc_auc',
    cv=cv,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print('Mejores hiperparámetros encontrados:')
print(grid_search.best_params_)
print(f'Mejor AUC-ROC en validación cruzada: {grid_search.best_score_:.4f}')
```

```
# Guardamos el mejor modelo encontrado para la evaluación final
mejor_modelo = grid_search.best_estimator_
```

## 7. Evaluación final en el conjunto de prueba

```
# Predicciones sobre el conjunto de prueba (nunca antes visto por el modelo)
y_pred = mejor_modelo.predict(X_test)
y_pred_proba = mejor_modelo.predict_proba(X_test)[:, 1]  # probabilidad de la clase 1 (churn)

# Métrica principal: AUC-ROC
auc_roc_final = roc_auc_score(y_test, y_pred_proba)

# Métrica adicional: exactitud
exactitud_final = accuracy_score(y_test, y_pred)

print(f'AUC-ROC en test : {auc_roc_final:.4f}')
print(f'Exactitud en test: {exactitud_final:.4f}')

# Ubicamos el resultado dentro de la tabla de puntuación del proyecto
if auc_roc_final < 0.75:
    sp = 0
elif auc_roc_final < 0.81:
    sp = 4
elif auc_roc_final < 0.85:
    sp = 4.5
elif auc_roc_final < 0.87:
    sp = 5
elif auc_roc_final < 0.88:
    sp = 5.5
else:
    sp = 6

print(f'Puntuación estimada del proyecto (SP): {sp}')
```

```
# Curva ROC
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)

plt.figure(figsize=(6, 6))
plt.plot(fpr, tpr, label=f'AUC-ROC = {auc_roc_final:.4f}', color='#DD8452')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Modelo aleatorio')
plt.xlabel('Tasa de falsos positivos')
plt.ylabel('Tasa de verdaderos positivos')
plt.title('Curva ROC - Modelo final')
plt.legend()
plt.show()
```

```
# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Activo', 'Cancelado'])
disp.plot(cmap='Blues')
plt.title('Matriz de confusión - Modelo final')
plt.show()
```

```
# Importancia de características del modelo final (útil para el informe de negocio)
importancias = pd.Series(mejor_modelo.feature_importances_, index=X_train.columns)
importancias = importancias.sort_values(ascending=False).head(15)

plt.figure(figsize=(8, 6))
importancias.plot(kind='barh', color='#4C72B0')
plt.gca().invert_yaxis()
plt.xlabel('Importancia')
plt.title('Top 15 variables más importantes para predecir churn')
plt.show()
```

## 8. Conclusiones

*(Completar tras ejecutar el notebook con los datos reales; esta sección resume el tipo
de hallazgos que se esperan en base al análisis y sirve de base para el informe final.)*

**Sobre los datos:**
- El dataset presenta un desbalance de clases moderado (~26-27% de cancelaciones), lo cual
  se abordó mediante `class_weight`/`scale_pos_weight` en lugar de sobre/submuestreo agresivo.
- Los nulos estructurales (clientes sin internet o teléfono) se trataron como categoría
  explícita en vez de eliminarse, preservando información relevante.

**Sobre las variables predictoras:**
- La **antigüedad del cliente** y el **tipo de contrato** (mes a mes vs. 1-2 años) son
  consistentemente las variables más relevantes: los clientes nuevos con contrato mensual
  son el segmento de mayor riesgo de cancelación.
- El **método de pago** (cheque electrónico) y el **servicio de fibra óptica** también
  muestran una asociación relevante con el churn.

**Sobre el modelo:**
- El modelo final (LightGBM/CatBoost, según el mejor resultado obtenido) alcanzó un
  AUC-ROC de **`{{completar con el valor real de auc_roc_final}}`**, superando el umbral
  mínimo de 0.75 requerido por el proyecto.
- La exactitud por sí sola no es suficiente para evaluar este problema dado el desbalance
  de clases; por eso el AUC-ROC es la métrica principal de referencia.

**Limitaciones:**
- Los datos son un corte transversal a partir del 1 de febrero de 2020; el modelo no
  captura cambios de comportamiento posteriores a esa fecha.
- No se dispone de información sobre motivos de cancelación (encuestas de salida) que
  podrían enriquecer el análisis causal.

**Recomendación de negocio:**
- Priorizar campañas de retención (códigos promocionales, migración a contratos anuales)
  hacia clientes con **contrato mes a mes**, **poca antigüedad** y **pago por cheque
  electrónico**, ya que representan el segmento de mayor probabilidad de cancelación
  según el modelo.
