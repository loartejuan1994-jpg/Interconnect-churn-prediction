```python
# Importe de librerías
import pandas as pd
print('Librerías importadas exitosamente')
```

```python
# Carga de datasets

df_contract_orig = pd.read_csv('contract.csv')      # información del contrato
df_internet_orig = pd.read_csv('internet.csv')      # información sobre los servicios de Internet
df_personal_orig = pd.read_csv('personal.csv')      # datos personales del cliente
df_phone_orig    = pd.read_csv('phone.csv')         # información sobre los servicios telefónicos

# Generar copias
df_contract = df_contract_orig.copy()
df_internet = df_internet_orig.copy()
df_personal = df_personal_orig.copy()
df_phone    = df_phone_orig.copy()

# Generar diccionario
datasets = {'Contract' : df_contract,
            'Internet' : df_internet,
            'Personal' : df_personal,
            'Phone' : df_phone
           }

print("Datasets cargados correctamente")
```

```python
# Exploración inicial de datos
for nombre, df in datasets.items():

    print(f'\n{"="*60}')
    print(f" {nombre.upper()} ")
    print(f'{"="*60}')

    print(f"\n📌 PRIMERAS 5 FILAS ")
    display(df.head())

    print(f"\n📌 INFORMACIÓN GENERAL")
    df.info()

    print(f"\n📌 ESTADÍSTICA DESCRIPTIVA ")
    display(df.describe(include='all').T)

    print(f"\n📌 DIMENCIONES DE DATOS")
    print(f"Número de filas       : {df.shape[0]}")
    print(f"Número de columnas    : {df.shape[1]}")
    print(f"Valores nulos totales : {df.isnull().sum().sum()}")
    print(f"Valores duplicados    : {df.duplicated().sum()}")
    

    print("--" * 40)
```

```python
# Resúmen de los 4 datasets
resumen = pd.DataFrame({
    nombre: {
        'Filas': df.shape[0],
        'Columnas': df.shape[1],
        'Nulos': df.isna().sum().sum(),
        'Duplicados': df.duplicated().sum()
    }
    for nombre, df in datasets.items()
}).T

display(resumen)
```
